```python
import hashlib
```

```python
def hash_data(data):
    return hashlib.sha256(data.encode()).hexdigest()
```

```python
gym_clients = ["Yossi","Jeff","Miriam"]
library_clients = ["Yossi","Tammy","George"]
```

```python
hashed_gym_clients = [hash_data(name) for name in gym_clients]
hashed_library_clients = [hash_data(name) for name in library_clients]
```
